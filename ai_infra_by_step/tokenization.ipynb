# Tokenization

Ref: [Hugging Face - Tokenizers](https://huggingface.co/learn/llm-course/chapter2/4)

Goal: Present raw text as numbers in the most meaningful and smallest way, "unknown" token as few as possible

## Word-based

Benefits:

1. Easy to set up and use with only a few rules (may have extra rules for punctuation), often yields decent results

Limitation:

1. Large “vocabularies” (over 500,000 words in the English language)
2. Can't identify related word, “dog” vs “dogs” 

## Character-based

Benefits:

1. Smaller vocabulary
2. Much fewer out-of-vocabulary (unknown) tokens, since every word can be built from characters.

Limitation:

1. Less meaningful: each character doesn’t mean a lot on its own, whereas that is the case with words. This differs according to the language; in Chinese, for example, each character carries more information than a character in a Latin language.

### Subword-based

Principle: frequently used words should not be split into smaller subwords, but rare words should be decomposed into meaningful subwords.

Benefits:

1. Relatively good coverage with small vocabularies, and close to no unknown tokens.
2. Especially useful in agglutinative languages such as Turkish, where you can form (almost) arbitrarily long complex words by stringing together subwords.

## Others

1. Byte-level BPE, as used in GPT-2
2. WordPiece, as used in BERT
3. SentencePiece or Unigram, as used in several multilingual models

## Encoding

Translating text to numbers. 2 steps:

1. Split the text into words (or parts of words, punctuation symbols, etc.), usually called *tokens*.
2. Convert those tokens into numbers

```python
from transformers import BertTokenizer, AutoTokenizer

sequence1 = "Using a Transformer network is simple"
sequence2 = "试一下中文"
tokenizer = BertTokenizer.from_pretrained("bert-base-cased") # download the vocabulary
```

```python
print(tokenizer(sequence1))
print(tokenizer(sequence2))

# Encoding: convert input text to token IDs
```

```python
# 2 steps separately

# Step 1
tokens1 = tokenizer.tokenize(sequence1)
print(tokens1)
# Step 2
ids1 = tokenizer.convert_tokens_to_ids(tokens1)
print(ids1)

# Step 1
tokens2 = tokenizer.tokenize(sequence2)
print(tokens2)
# Step 2
ids2 = tokenizer.convert_tokens_to_ids(tokens2)
print(ids2)
```

## Decoding

From vocabulary indices to string

```python
decoded_string = tokenizer.decode([7993, 170, 11303, 1200, 2443, 1110, 3014])
print(decoded_string)
```

```python
decoded_string = tokenizer.decode([100, 976, 979, 980, 1030])
print(decoded_string)
```
